# Pilot Analysis — Groq Llama-3.3-70B
Histogram + descriptive stats cho pilot (30 method × 2 điều kiện = 60 lần gọi).
Nguồn: `results/pilot_llm_output.csv` (rút từ `PILOT_RESULTS.md`).

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('pilot_llm_output.csv')
print('Total runs:', len(df))
df.head()
```

```python
# Descriptive statistics
print(df[['latency_s','tokens_per_s']].describe())
print('\nTheo điều kiện (C1 zero-shot vs C2 few-shot):')
print(df.groupby('condition')[['latency_s','tokens_per_s']].mean())
```

```python
# Histograms
fig, ax = plt.subplots(1, 2, figsize=(11,4))
ax[0].hist(df['latency_s'], bins=12, color='#4C78A8', edgecolor='white')
ax[0].set_title('Latency (s)'); ax[0].set_xlabel('s'); ax[0].set_ylabel('count')
ax[1].hist(df['tokens_per_s'], bins=12, color='#F58518', edgecolor='white')
ax[1].set_title('Throughput (tokens/s)'); ax[1].set_xlabel('tokens/s')
plt.tight_layout(); plt.show()
```

## Nhận xét
- Latency trung bình ~0.78s (0.29–1.43s), throughput ~538 tokens/s (310–720).
- Không gặp rate-limit / lỗi server trong pilot.
- Pilot này validate toàn bộ pipeline trên 1 model (Llama-3.3-70B) trước khi chạy 3 model còn lại.
